# Brightness debug notebook

Run the setup cell first, then run the extraction cell once. Use the dropdowns below to inspect one condition, one marker, and optionally one donor at a time without stacking every debug plot in the notebook.

```python
import os
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / "macrophage_analysis").exists():
    raise RuntimeError("Run this notebook from the repository root or notebooks/ directory.")
os.chdir(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
import ipywidgets as widgets
from IPython.display import display
import macrophage_analysis as ma

debug_donors = list(ma.DEFAULT_DONORS)
debug_conditions = list(ma.DEFAULT_CONDITIONS)
debug_antibody_order = list(ma.DEFAULT_ANTIBODY_ORDER)
debug_donors, debug_conditions, debug_antibody_order
```

Run this once to calculate the per-image brightness values. The widget cell below reuses `debug_results` and only reloads the original images needed for the selected condition, marker, and donor.

```python
debug_results = ma.extract_single_cell_fluorescence(
    donors=debug_donors,
    conditions=debug_conditions,
    antibody_order=debug_antibody_order,
)
```

```python
def plot_selected_brightness_debug(
    condition: str,
    marker: str,
    donor: str = "__all__",
) -> None:
    selected_donors = None if donor == "__all__" else [donor]
    ma.plot_condition_brightness_debug(
        debug_results,
        condition,
        donors=selected_donors,
        antibody_order=[marker],
        figure_width_per_marker=14.0,
        figure_height_per_donor=7.5,
    )

condition_dropdown = widgets.Dropdown(
    options=debug_conditions,
    value=debug_conditions[0],
    description="Condition",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="320px"),
)

marker_dropdown = widgets.Dropdown(
    options=debug_antibody_order,
    value=debug_antibody_order[0],
    description="Protein / marker",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="320px"),
)

donor_dropdown = widgets.Dropdown(
    options=[("All donors", "__all__")] + [(donor, donor) for donor in debug_donors],
    value="__all__",
    description="Donor (optional)",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="320px"),
)

controls = widgets.VBox([condition_dropdown, marker_dropdown, donor_dropdown])
output = widgets.interactive_output(
    plot_selected_brightness_debug,
    {
        "condition": condition_dropdown,
        "marker": marker_dropdown,
        "donor": donor_dropdown,
    },
)

display(controls, output)
```
