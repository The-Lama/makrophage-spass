# Relative brightness analysis

This notebook compares each cell's marker signal to the background lighting of its own image. The metric is `cell mean intensity / median non-cell background intensity`, using StarDist labels to identify non-cell background pixels.

```python
import os

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

import macrophage_analysis as ma

relative_donors = list(ma.DEFAULT_DONORS)
relative_conditions = list(ma.DEFAULT_CONDITIONS)
relative_antibody_order = list(ma.DEFAULT_ANTIBODY_ORDER)
relative_n_tiles = (4, 4)
relative_background_percentile = 50.0
relative_plot_log_scale = False
relative_plot_upper_percentile = 99.0

relative_donors, relative_conditions, relative_antibody_order
```

Run this once. If your GPU still runs out of memory, restart the kernel, change `relative_n_tiles` to `(8, 8)` in the setup cell, and run from the setup cell again.

```python
relative_results = ma.extract_single_cell_relative_fluorescence(
    donors=relative_donors,
    conditions=relative_conditions,
    antibody_order=relative_antibody_order,
    n_tiles=relative_n_tiles,
    background_percentile=relative_background_percentile,
)
```

These plots use signal/background ratio instead of raw brightness.

```python
relative_stats = ma.plot_violins_with_stats(
    relative_results,
    plot_log_scale=relative_plot_log_scale,
    upper_display_percentile=relative_plot_upper_percentile,
    y_label="Cell signal / image background",
    title_value_label="signal/background ratio",
)
ma.plot_summary_heatmap(
    relative_results,
    stat_results=relative_stats,
    summary_stat_label="median signal/background ratio",
)
ma.display_stat_summary_tables(
    relative_stats,
    baseline_condition=relative_results.baseline_condition,
)
```
